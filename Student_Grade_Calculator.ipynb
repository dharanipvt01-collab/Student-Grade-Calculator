# Student Grade Calculator

This project calculates student grades based on their marks
using predefined grading conditions

## Grading Criteria

| Marks | Grade |
|---|---|
| 90–100 | A+ |
| 80–89 | A |
| 70–79 | B |
| 60–69 | C |
| 50–59 | D |
| 40–49 | E |
| 0–39 | F |

```python
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid"
    elif marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    elif marks >= 40:
        return "E"
    else:
        return "F"
```

```python
students = [
    ("Dharani", 95),
    ("Arun", 87),
    ("Priya", 76),
    ("Rahul", 64),
    ("Anu", 52),
    ("Kiran", 38)
]
```

```python
for name, marks in students:
    grade = calculate_grade(marks)
    print(f"{name:<10} Marks: {marks:<5} Grade: {grade}")
```

## Testing

The program is tested using different marks to verify that each grading condition works correctly.

```python
test_marks = [95, 87, 76, 64, 52, 45, 38, 105, -5]

for marks in test_marks:
    print(f"Marks: {marks:<5} Grade: {calculate_grade(marks)}")
```

## Conclusion

The Student Grade Calculator successfully assigns grades according to predefined marks ranges and identifies invalid marks.
